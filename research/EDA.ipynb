```python
kaggle_dataset = pd.read_csv("../data/raw/gold-dataset-sinha-khandait.csv")
kaggle_dataset.head()
```

```python
y_finance_dataset = pd.read_csv("../data/raw/YFinance_data.csv")
y_finance_dataset.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

kaggle_dataset["Dates"] = pd.to_datetime(
    kaggle_dataset["Dates"], dayfirst=True, errors="coerce"
)
y_finance_dataset["Date"] = pd.to_datetime(y_finance_dataset["Date"])

y_finance_dataset.set_index("Date", inplace=True)

headline_dates = kaggle_dataset["Dates"]

plt.figure(figsize=(14, 7))
plt.plot(
    y_finance_dataset.index, y_finance_dataset["Open"], label="Open Price", color="blue"
)

headline_data = y_finance_dataset.loc[y_finance_dataset.index.isin(headline_dates)]
plt.scatter(
    headline_data.index,
    headline_data["Open"],
    color="red",
    label="Headline Dates",
    zorder=5,
)

plt.title("Gold Open Price Over Time with Headlines Highlighted")
plt.xlabel("Date")
plt.ylabel("Open Price")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = " ".join(kaggle_dataset["News"].dropna())

wordcloud = WordCloud(
    width=1600, height=800, background_color="white", colormap="viridis", max_words=200
).generate(text)

plt.figure(figsize=(16, 8))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Gold Headlines", fontsize=20)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

kaggle_dataset["Dates"] = pd.to_datetime(
    kaggle_dataset["Dates"], dayfirst=True, errors="coerce"
)
kaggle_dataset = kaggle_dataset.dropna(subset=["Dates"])

kaggle_dataset.set_index("Dates", inplace=True)

headline_counts = kaggle_dataset["News"].resample("M").count()

plt.figure(figsize=(14, 7))
plt.plot(
    headline_counts.index,
    headline_counts.values,
    marker="o",
    linestyle="-",
    color="purple",
)
plt.title("Number of Gold Headlines Over Time", fontsize=18)
plt.xlabel("Date", fontsize=14)
plt.ylabel("Number of Headlines", fontsize=14)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the groups
baseline = np.array([0.4811, 0.4777, 0.4845, 0.4914, 0.4845, 0.4770])
w_sentiment = np.array([0.5086, 0.4914, 0.5085, 0.5017, 0.4914, 0.4879])
w_cluster = np.array([0.4983, 0.5052, 0.5017, 0.5258, 0.4948, 0.5258])
w_combined = np.array([0.5223, 0.5258, 0.4982, 0.5051, 0.5326, 0.5120])

# Calculate means
means = [
    np.mean(baseline),
    np.mean(w_sentiment),
    np.mean(w_cluster),
    np.mean(w_combined),
]

# Calculate standard deviations
stds = [
    np.std(baseline, ddof=1),
    np.std(w_sentiment, ddof=1),
    np.std(w_cluster, ddof=1),
    np.std(w_combined, ddof=1),
]

# Labels
labels = ["Baseline", "w Sentiment", "w Cluster", "w Combined"]

# Plot
fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.bar(
    labels,
    means,
    yerr=stds,
    capsize=10,
    color=["skyblue", "lightgreen", "salmon", "orchid"],
)

# Labels and title
ax.set_ylabel("Score")
ax.set_title("Performance Comparison with Error Bars (Std Dev)")
ax.set_ylim(0.45, 0.55)  # Adjust if needed

# Show values on top of bars
for bar, mean in zip(bars, means):
    height = bar.get_height()
    ax.annotate(
        f"{mean:.3f}",
        xy=(bar.get_x() + bar.get_width() / 2, height),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center",
        va="bottom",
    )

plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
baseline = np.array([0.4811, 0.4777, 0.4845, 0.4914, 0.4845, 0.4770])
w_sentiment = np.array([0.5086, 0.4914, 0.5085, 0.5017, 0.4914, 0.4879])
w_cluster = np.array([0.4983, 0.5052, 0.5017, 0.5258, 0.4948, 0.5258])
w_combined = np.array([0.5223, 0.5258, 0.4982, 0.5051, 0.5326, 0.5120])

means = [
    np.mean(baseline),
    np.mean(w_sentiment),
    np.mean(w_cluster),
    np.mean(w_combined),
]

stds = [
    np.std(baseline, ddof=1),
    np.std(w_sentiment, ddof=1),
    np.std(w_cluster, ddof=1),
    np.std(w_combined, ddof=1),
]

labels = [
    "Baseline",
    "Basline with Sentiment",
    "Basline with Cluster",
    "Basline with both",
]

# Colors – highlight last bar
colors = ["#BFBFBF", "#BFBFBF", "#BFBFBF", "#8E44AD"]  # grey, grey, grey, purple

sorted_idx = np.argsort(means)
sorted_idx = np.argsort(means)[::-1]

means = np.array(means)[sorted_idx]
stds = np.array(stds)[sorted_idx]
labels = np.array(labels)[sorted_idx]
colors = np.array(colors)[sorted_idx]

fig, ax = plt.subplots(figsize=(9, 6))

bars = ax.barh(
    labels,
    means,
    capsize=8,
    error_kw=dict(ecolor="dimgray", elinewidth=2, alpha=0.8, capsize=8),
    color=colors,
)

ax.set_title(
    "Cluster insights and sentiment cues sharpen prediction quality, with the combined approach outperforming all baselines.",
    fontsize=15,
    fontweight="bold",
    pad=15,
)

ax.set_xlabel("Average Score (with Uncertainty)", fontweight="bold")


ax.set_xlim(0.45, 0.56)

for bar, mean, std in zip(bars, means, stds):
    x = mean + 0.002
    ax.annotate(
        f"{mean:.3f}",
        xy=(x, bar.get_y() + bar.get_height() / 2),
        xytext=(3, 0),
        textcoords="offset points",
        va="center",
        fontsize=11,
        fontweight="semibold",
    )

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
